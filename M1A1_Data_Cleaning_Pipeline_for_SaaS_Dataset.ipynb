# Task 4 – Data Cleaning Pipeline

Create functions that perform the following cleaning steps.

1. Handle Missing Values

Rules:

● If the name is missing → replace it with "Unknown Customer."


● If usage_hours is NULL or missing → replace with 0

● If monthly_fee is invalid → mark it as None



2. Remove Duplicate Records

● Duplicate customers may appear in the dataset.

● Use customer_id to detect duplicates.

● Create a function: remove_duplicates(data), which removes the duplicates.



3. Standardize Date Formats

● Convert signup_date to the standard format: YYYY-MM-DD

● Use the datetime module to parse dates.

```python
# new data create within duplicates
dirty_data = {
    "customer_id": [101, 102, 103, 104, 101, 106, 107],

    "name": ["Rahul Sharma", "Meena Nair", None, "A. Kumar", "David", "Priya",
             "Rahul Sharma"],

    "signup_date": ["2023-01-10", "2023/02/15", "03-01-2023", "15-04-2023", "2023-05-10",
    "20230520", "2023-01-10"],

    "plan": ["Pro", "Basic", "Pro", "Enterprise", "Pro", "", "Pro"],

    "monthly_fee": ["49.99", "19.99", "forty-nine", "99.99", "49.99", "49.99", "49.99"],

    "usage_hours": ["120", "45", "88", None, "NULL", "200", "120"]
    }

print(dirty_data)
```

```python
from datetime import datetime


# 1. Handle Missing Values

def handle_missing_values(dirty_data):

    for i in range(len(dirty_data["customer_id"])):

        # Missing name
        if dirty_data["name"][i] is None or dirty_data["name"][i] == "":
            dirty_data["name"][i] = "Unknown Customer"

        # Missing usage hours
        if dirty_data["usage_hours"][i] is None or str(dirty_data["usage_hours"][i]).upper() == "NULL":
            dirty_data["usage_hours"][i] = 0

        # Invalid monthly fee
        try:
            dirty_data["monthly_fee"][i] = float(dirty_data["monthly_fee"][i])
        except (ValueError, TypeError):
            dirty_data["monthly_fee"][i] = None

    return dirty_data


# 2. Remove Duplicate Records

def remove_duplicates(dirty_data):

    seen_ids = set()
    duplicate_indexes = []

    for i in range(len(dirty_data["customer_id"])):

        customer_id = dirty_data["customer_id"][i]

        if customer_id in seen_ids:
            duplicate_indexes.append(i)
        else:
            seen_ids.add(customer_id)

    # Remove duplicate records
    for i in reversed(duplicate_indexes):

        for key in dirty_data:
            dirty_data[key].pop(i)

    return dirty_data


# 3. Standardize Date Formats

def standardize_dates(dirty_data):

    date_formats = [
        "%Y-%m-%d",
        "%Y/%m/%d",
        "%d-%m-%Y",
        "%Y%m%d"
    ]

    for i in range(len(dirty_data["signup_date"])):

        date_value = dirty_data["signup_date"][i]

        for fmt in date_formats:

            try:
                date_object = datetime.strptime(date_value, fmt)

                dirty_data["signup_date"][i] = date_object.strftime("%Y-%m-%d")

                break

            except ValueError:
                continue

    return dirty_data


# Apply all cleaning functions

dirty_data = handle_missing_values(dirty_data)

dirty_data = remove_duplicates(dirty_data)

dirty_data = standardize_dates(dirty_data)

print("\nCleaned Data:")
print(dirty_data)
```

Example Output

```python
# Here i used cleaned_data of saas_data
def find_low_usage_customers(cleaned_data):

    print("Low Usage Customers")

    for i in range(len(cleaned_data["customer_id"])):

        usage = int(cleaned_data["usage_hours"][i])

        if usage < 50:
            print(
                f"ID: {cleaned_data['customer_id'][i]} | "
                f"Name: {cleaned_data['name'][i]} | "
                f"Plan: {cleaned_data['plan'][i]} | "
                f"Usage Hours: {usage}"
            )

find_low_usage_customers(cleaned_data)
```

# Task 8 – Save Cleaned Data to File

● Save the cleaned dataset to a CSV file and name it saas_customers_cleaned.csv.

● Each row should contain: customer_id,name,signup_date,plan,monthly_fee,usage_hours

Example File Content

101, Rahul Sharma, 2023-01-10, Pro, 49.99, 120

● Use try, except, else, and finally to handle file errors.

● Also, create another file named plan_summary.txt.

Example Content

```python
import csv

# Save cleaned data to CSV file
try:
    with open("saas_customers_cleaned.csv", "w", newline="") as file:

        writer = csv.writer(file)

        writer.writerow([
            "customer_id",
            "name",
            "signup_date",
            "plan",
            "monthly_fee",
            "usage_hours"
        ])

        for i in range(len(cleaned_data["customer_id"])):
            writer.writerow([
                cleaned_data["customer_id"][i],
                cleaned_data["name"][i],
                cleaned_data["signup_date"][i],
                cleaned_data["plan"][i],
                cleaned_data["monthly_fee"][i],
                cleaned_data["usage_hours"][i]
            ])

except Exception as e:
    print("Error:", e)

else:
    print("CSV file created successfully.")

finally:
    print("CSV file operation completed.")


# Create plan_summary.txt
try:
    with open("plan_summary.txt", "w") as file:

        plans = {}

        for plan in cleaned_data["plan"]:
            plans[plan] = plans.get(plan, 0) + 1

        for plan, count in plans.items():
            file.write(f"{plan} : {count}\n\n")

except Exception as e:
    print("Error:", e)

else:
    print("Plan summary file created successfully.")

finally:
    print("Plan summary file operation completed.")

# Print CSV file content
print("\n--- saas_customers_cleaned.csv ---\n")

with open("saas_customers_cleaned.csv", "r") as file:
    print(file.read())


# Print plan summary content
print("--- plan_summary.txt ---\n")

with open("plan_summary.txt", "r") as file:
    print(file.read())
```
